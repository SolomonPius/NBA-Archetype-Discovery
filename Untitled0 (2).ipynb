1. Ensure that data is loaded correctly

```python
import pandas as pd

df = pd.read_excel("NBA_data.xlsx")
df.head()
```

2. Filter data, so that players with under 4 minutes a playing time per game (less data) do not get factored into the mdoel

```python
min_mp = 4
cluster_number = 10
learning_rate = 0.0005
epochs = 120
weight_d = 1e-4
df_filtered = df[df["MP"] >= min_mp].copy()

print("Original shape:", df.shape)
print("Filtered shape:", df_filtered.shape)
print(df_filtered.head())
```

```python
import random
import numpy as np
import torch

seed = 42
random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)
torch.cuda.manual_seed_all(seed)
```

3. Autoencoder and Clustering

```python
import numpy as np

from sklearn.preprocessing import StandardScaler

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

from sklearn.cluster import KMeans

df = df_filtered

# save player names
player_names = df["Player"]

# remove player column
X = df.drop(columns=["Player"]).copy()

# force numeric values
X = X.apply(pd.to_numeric, errors="coerce")

# replace missing values with column median
X = X.fillna(X.median())

# convert to numpy
X = X.values

# standardize data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# convert to pytorch tensors
X_tensor = torch.tensor(X_scaled, dtype=torch.float32)

dataset = TensorDataset(X_tensor)
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

# build autoencoder
class Autoencoder(nn.Module):

    def __init__(self, input_dim):
        super(Autoencoder, self).__init__()

        # encoder
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(0.1),
            nn.Linear(64, 32),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Dropout(0.1),
            nn.Linear(32, 8)
        )

        # decoder
        self.decoder = nn.Sequential(
            nn.Linear(8, 32),
            nn.ReLU(),
            nn.Linear(32, 64),
            nn.ReLU(),
            nn.Linear(64, input_dim)
        )

    def forward(self, x):
        z = self.encoder(x)
        x_recon = self.decoder(z)
        return x_recon
  # initialize model
input_dim = X_scaled.shape[1]

model = Autoencoder(input_dim)

criterion = nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate, weight_decay=weight_d)

# train

for epoch in range(epochs):

    total_loss = 0

    for batch in dataloader:

        x = batch[0]

        recon = model(x)

        loss = criterion(recon, x)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {total_loss:.4f}")

# Extract Embeddings
with torch.no_grad():
    embeddings = model.encoder(X_tensor).numpy()

# clustering

with torch.no_grad():
    embeddings = model.encoder(X_tensor).numpy()

kmeans = KMeans(n_clusters=cluster_number, random_state=42)
clusters = kmeans.fit_predict(embeddings)

results = pd.DataFrame({
    "Player": player_names,
    "Cluster": clusters
})
```

4. Get cluster sizes, and a representative player of each cluster

```python

# amount of players in each cluster
cluster_sizes = results["Cluster"].value_counts().sort_index()

# merge cluster labels with stats
df_filtered = df[df["MP"] >= min_mp].copy().reset_index(drop=True)
df_filtered["Cluster"] = clusters


# player representation
centers = kmeans.cluster_centers_
representatives = []

for i in range(kmeans.n_clusters):

    # indices of players in this cluster
    cluster_indices = np.where(clusters == i)[0]

    # embeddings of those players
    cluster_embeddings = embeddings[cluster_indices]

    # cluster center
    center = centers[i]

    # compute distances to center
    distances = np.linalg.norm(cluster_embeddings - center, axis=1)

    # find closest player
    closest_index = cluster_indices[np.argmin(distances)]

    representatives.append(player_names.iloc[closest_index])

# combine cluster size + representative player
rep_df = pd.DataFrame({
    "Cluster": range(kmeans.n_clusters),
    "Cluster Size": cluster_sizes.values,
    "Representative Player": representatives
})

print(rep_df)

```

5. Get data from each cluster. Cluster size, representative player, 3 example players and average stats

```python


# make sure this matches the exact data used for training/clustering
model_df = df[df["MP"] >= min_mp].copy().reset_index(drop=True)

# optional pruning step if you used it
stat_df = model_df.drop(columns=["Player"]).copy()
stat_df = stat_df.apply(pd.to_numeric, errors="coerce")

# attach clusters
model_df["Cluster"] = clusters

# all numeric stats
numeric_cols = model_df.drop(columns=["Player", "Cluster"]).columns
cluster_summary = model_df.groupby("Cluster")[numeric_cols].mean()

# representative players from rep_df
rep_map = dict(zip(rep_df["Cluster"], rep_df["Representative Player"]))

def get_closest_players_in_cluster(cluster_id, n_players=5):
    cluster_indices = np.where(clusters == cluster_id)[0]
    cluster_embeddings = embeddings[cluster_indices]
    center = kmeans.cluster_centers_[cluster_id]
    distances = np.linalg.norm(cluster_embeddings - center, axis=1)
    sorted_idx = cluster_indices[np.argsort(distances)]
    return player_names.iloc[sorted_idx[:n_players]].tolist()

def format_cluster_for_ai(cluster_id, n_examples=5):
    size = int((clusters == cluster_id).sum())
    rep_player = rep_map.get(cluster_id, "Unknown")
    example_players = get_closest_players_in_cluster(cluster_id, n_examples)
    avg_stats = cluster_summary.loc[cluster_id]

    lines = []
    lines.append(f"Cluster {cluster_id}")
    lines.append(f"Size: {size}")
    lines.append(f"Representative player: {rep_player}")
    lines.append(f"Example players: {', '.join(example_players)}")
    lines.append("Average stats:")

    for stat, value in avg_stats.items():
        lines.append(f"- {stat}: {value:.3f}")



    return "\n".join(lines)

# print all clusters
for cluster_id in sorted(model_df["Cluster"].unique()):
    print("=" * 80)
    print(format_cluster_for_ai(cluster_id, n_examples=5))
    print()
```

6. Find out what cluster any given player is in

```python
def find_player_cluster(player_name):
    match = results[results["Player"].str.contains(player_name, case=False, na=False)]

    if len(match) == 0:
        return "Player not found."

    if len(match) > 1:
        return match[["Player", "Cluster"]]

    player = match.iloc[0]["Player"]
    cluster = match.iloc[0]["Cluster"]

    return f"{player} is in Cluster {cluster}."

print(find_player_cluster("LeBron James"))
print(find_player_cluster("nikola jokic"))
print(find_player_cluster("stephen curry"))
```

7. List all players in a cluster, ordered by a certain statistic

```python
def list_players_in_cluster(cluster_id, sort_by="MP", ascending=False, cols=None):
    cluster_df = model_df[model_df["Cluster"] == cluster_id].copy()
    cluster_df = cluster_df.sort_values(by=sort_by, ascending=ascending)

    if cols is None:
        cols = ["Player", sort_by]

    return cluster_df[cols]

pd.set_option("display.max_rows", None)
print(list_players_in_cluster(0, sort_by="MP", cols=["Player", "MP", "FGA", "USG%"]))
```

8. Given some new player data, what cluster would they belong to

```python
X_columns = model_df.drop(columns=["Player", "Cluster"]).columns
X_medians = model_df.drop(columns=["Player", "Cluster"]).apply(pd.to_numeric, errors="coerce").median()

def predict_new_player_cluster(new_player_stats):
    # new_player_stats should be a dictionary:
    # {"shots in 0-3": ..., "shots in 3-10": ..., ..., "FGA": ...}

    # make a 1-row dataframe
    new_df = pd.DataFrame([new_player_stats])

    # make sure column order matches training data
    new_df = new_df[X_columns]

    # force numeric
    new_df = new_df.apply(pd.to_numeric, errors="coerce")

    # fill any missing values with the training medians
    new_df = new_df.fillna(X_medians)

    # apply the SAME scaler used on training data
    new_scaled = scaler.transform(new_df.values)

    # convert to tensor
    new_tensor = torch.tensor(new_scaled, dtype=torch.float32)

    # Set model to evaluation mode
    model.eval()

    # get embedding
    with torch.no_grad():
        new_embedding = model.encoder(new_tensor).numpy()

    # Set model back to training mode (if subsequent training is expected, though not critical here)
    model.train()

    # predict cluster
    predicted_cluster = kmeans.predict(new_embedding)[0]

    return predicted_cluster

new_player = {
    "shots in 0-3": 0.28,
    "shots in 3-10": 0.18,
    "shots in 10-16": 0.04,
    "shots in 16-3P": 0.02,
    "shots in 3P": 0.48,
    "2P shot assisted": 0.72,
    "3P shot assisted": 0.96,
    "%FGA": 0.08,
    "%3PA": 0.52,
    "TS%": 0.58,
    "3PAr": 0.51,
    "FTr": 0.21,
    "ORB%": 4.5,
    "DRB%": 12.8,
    "AST%": 9.3,
    "STL%": 1.8,
    "BLK%": 1.2,
    "TOV%": 10.4,
    "USG%": 16.2,
    "MP": 28.0,
    "FGA": 8.7
}

cluster = predict_new_player_cluster(new_player)
print("Predicted cluster:", cluster)
```

9. PCA (Baseline Model)

```python
from sklearn.decomposition import PCA

from sklearn.metrics import silhouette_score

from sklearn.cluster import KMeans


pca = PCA(n_components=8)
X_pca = pca.fit_transform(X_scaled)


kmeans = KMeans(n_clusters=cluster_number, random_state=42)
clusters = kmeans.fit_predict(X_pca)


score = silhouette_score(X_pca, clusters)
print("Baseline silhouette score:", score)

with torch.no_grad():
    embeddings = model.encoder(X_tensor).numpy()

kmeans = KMeans(n_clusters=cluster_number, random_state=42, n_init=10)
clusters = kmeans.fit_predict(embeddings)

score = silhouette_score(embeddings, clusters)
print("Model silhouette score:", score)
```
